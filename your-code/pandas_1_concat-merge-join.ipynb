## Concatenating

Concatenating two dataframes combines two dataframes so that we append the rows of one dataframe at the end of the other. Our column names have to be identical for this function to work properly.

Below is an example of the `concat()` function in pandas

https://pandas.pydata.org/docs/reference/api/pandas.concat.html

```python
df1 = pd.DataFrame({'A': ['a'+str(x) for x in range(3)],
                    'B': ['b'+str(x) for x in range(3)],
                    'C': ['c'+str(x) for x in range(3)]},
                     index=[0, 1, 2])

df2 = pd.DataFrame({'A': ['a'+str(x) for x in range(3, 6)],
                    'B': ['b'+str(x) for x in range(3, 6)],
                    'C': ['c'+str(x) for x in range(3, 6)]},
                     index=[3, 4, 5]) 

df3 = pd.DataFrame({'D': ['d'+str(x) for x in range(3)],
                    'E': ['e'+str(x) for x in range(3)],
                    'F': ['f'+str(x) for x in range(3)]},
                     index=[0, 1, 2]) 

df4 = pd.DataFrame({'D': ['d'+str(x) for x in range(3, 6)],
                    'E': ['e'+str(x) for x in range(3, 6)],
                    'F': ['f'+str(x) for x in range(3, 6)]},
                     index=[3, 4, 5]) 

# print(df1, '\n---\n', df2, '\n---\n', df3, '\n---\n',df4)
```

Let's try concatenating `df1` and `df2`, as well as `df3` and `df4`. 

```python
# your code here
# df1 and df2 share columns A, B, C, and their indexes do not overlap
# (0-2 then 3-5), so stacking them simply continues the rows.
print("concat df1 and df2:")
display(pd.concat([df1, df2]))

# df3 and df4 behave the same way with columns D, E, F
print("concat df3 and df4:")
display(pd.concat([df3, df4]))
```

From the output above, you see the second dataframe is appended at the bottom of the first dataframe.

Now let's try concatenating `df1`, `df2`, `df3`, and `df4` all together.

Note that the `sort=False` param is supplied to silence a warning message on a future Pandas change. It does not make any difference on the output.

```python
# your code here
pd.concat([df1, df2, df3, df4], sort=False)

# 12 rows and 6 columns, more than half of it NaN.
#
# Two things are happening:
#
# 1. concat defaults to axis=0, so it stacks ROWS. df3 and df4 bring column
#    names that df1 and df2 do not have, so the frame has to widen and the
#    gaps are filled with NaN.
#
# 2. The index values 0-5 now appear TWICE - once from df1/df2 and once from
#    df3/df4 - because concat preserves the original indexes. df.loc[0] would
#    return two rows.
```

What do we find?

* Pandas' `concat` method respects indexes of all axes. 
    * Because `df3` and `df4` have different column indexes than `df1` and `df2`, `concat` put them into different columns. 
    * `df3` and `df4` also retain their original row indexes of 0-5 instead of continuing from the last index of `df2`. 
* `concat` creates `NaN` at places where values are missing.

Try also supplying `ignore_index=True` to `concat`. How is the output different?

```python
# your code here
pd.concat([df1, df2, df3, df4], ignore_index=True, sort=False)

# The only difference: the index runs 0-11 instead of 0-5 twice. The original
# labels are discarded and the rows are numbered fresh.
#
# What ignore_index does NOT fix is the block of NaN. It only relabels rows -
# it cannot align df3's D, E, F onto df1's rows, because stacking vertically
# was never going to do that.
#
# That is the clue for the bonus question at the end: to get a 6x6 frame with
# no NaN, the data has to be joined side by side (axis=1), not stacked.
```

## Merging and Joining

Pandas has two functions for joining datasets: `merge()` and `join()`. They perform the same task but have different options and syntax. 

Below is an example of `merge` and `join`.     
HINT (uses the column that repeats in both dataframes )

https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.merge.html

```python
left = pd.DataFrame({'idx': ['i'+str(x) for x in range(3)],
                     'A': ['a'+str(x) for x in range(3)],
                     'B': ['b'+str(x) for x in range(3)]})


right = pd.DataFrame({'idx': ['i'+str(x) for x in range(1,4)],
                     'C': ['c'+str(x) for x in range(1,4)],
                     'D': ['d'+str(x) for x in range(1,4)]})
```

```python
left
```

```python
right
```

`join` is identical to `merge`. But when using join, we need to explicitly set the index column of the dataframes to join using `set_index`:

https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.join.html

```python
# your code here
# join works on the INDEX, so the column we want to join on has to be moved
# into the index first with set_index.
joined = left.set_index('idx').join(right.set_index('idx'))
display(joined)

# The same result with merge, which takes the column name directly:
merged = left.merge(right, on='idx', how='left')
display(merged)

# The markdown above says join is identical to merge. Almost - but their
# DEFAULTS differ, and that is worth knowing:
#
#     join  defaults to how='left'    (keep every row of the left frame)
#     merge defaults to how='inner'   (keep only rows present in both)
#
# That is exactly why the join above keeps i0 with NaN in C and D, but
# drops i3 entirely. i0 exists only in left, i3 only in right.
#
# Compare all four options:
print("left  idx:", list(left['idx']))
print("right idx:", list(right['idx']))
print()
for how in ['inner', 'left', 'right', 'outer']:
    out = left.merge(right, on='idx', how=how)
    print("how=%-6s -> %d rows, idx: %s" % (how, len(out), list(out['idx'])))

# inner  keeps only i1, i2        - the overlap
# left   keeps i0, i1, i2         - everything in left
# right  keeps i1, i2, i3         - everything in right
# outer  keeps i0, i1, i2, i3     - everything from either side
```

And you see, `join` disregards the row of `right` with the unmatching index `i3`. It retains the row of `left` with the unmatching index `i0` but uses `NaN` for the missing data after joining.

#### There are other options we can explore with the `merge()` and `join()` functions. 

Specifically, we can specify `how`. This argument in the function tells us whether we are performing an inner, left, right, or outer join.

We can also specify a different column for joining in the `merge()` function using the `left_on` and `right_on` arguments. Check out the following documentations if you want to explore more:

[pandas.DataFrame.merge](https://pandas.pydata.org/pandas-docs/stable/generated/pandas.DataFrame.merge.html)

[pandas.DataFrame.join](https://pandas.pydata.org/pandas-docs/stable/generated/pandas.DataFrame.join.html)

Now if you look back on `merge` and `join`, you realize that in order to perform these functions on a set of dataframes, these dataframes must share a common column as the index. Only rows that have the same index values will be joined. This is similar to the [`join` function in MySQL](https://www.w3schools.com/sql/sql_join.asp), isn't it?

The bonus question for you is to figure out how to join and concatenate `df1`, `df2`, `df3`, and `df4` we created at the beginning of this challenge. Your end product should look like this:

```python
# your code here
# Two steps, in two different directions.
#
# Step 1 - CONCAT vertically. df1 and df2 share columns A, B, C, so stacking
# them gives one frame with rows 0-5. Same for df3 and df4 with D, E, F.
abc = pd.concat([df1, df2])      # rows 0-5, columns A B C
de_f = pd.concat([df3, df4])     # rows 0-5, columns D E F

# Step 2 - JOIN horizontally. Both frames now carry the same index (0-5),
# which is the shared key join needs. This lines row 0 up with row 0 instead
# of putting them one below the other.
result = abc.join(de_f)

# The target image labels the index axis "index"
result = result.rename_axis('index')

display(result)
print("shape:", result.shape, " NaN count:", int(result.isnull().sum().sum()))

# Equivalent one-liners, all producing the same frame:
#
#   pd.concat([abc, de_f], axis=1)
#   abc.merge(de_f, left_index=True, right_index=True)
#   pd.concat([pd.concat([df1, df2]), pd.concat([df3, df4])], axis=1)
#
# The point of the exercise is the direction. Cell 9 concatenated all four
# along axis=0 and produced 12 rows that were more than half NaN. The same
# four frames joined along axis=1 give 6 rows and no NaN at all.
#
# Nothing about the data changed - only whether we treated the shared index
# 0-5 as rows to stack or as a key to align on. That is the whole difference
# between concat and join: concat glues frames together, join matches them up.
```
